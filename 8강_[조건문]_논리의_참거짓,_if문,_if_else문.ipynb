**Example)**

```python
n = 90
if n >= 80 :
  print("시험에 합격하셨습니다.")
  print("축하합니다.")

print("프로그램을 종료합니다.")
```

*if문은 조건을 충족하지 않는다면 실행되지 않는다.

```python
if sum([1,2,3]) > 1 :
  print([1,2,3])
print("out of it")
```

```python
if 3 :
  print("three")

if 0 :
  print("zero")
```

*실제로 프로그래밍은 0은 False, 1은 True라고 인식함. 여기서는 0만 아니면 모두 True인 상태. 그렇기에 첫번째 if문은 인식되어 three라는 결과값을 출력하지만 두번째 if문은 그렇지 않음.

```python
if "three" :
  print("3")

if "" :
  print("0")
```

*문자열의 경우, ""안에 아무 것도 없는 경우를 False라고 인식.

```python
if " " :
  print("three")
```

*그러나 " "의 경우, space bar 한 칸도 하나의 문자열로 써 인식되므로 True.

```python
if [1,2,3] :
  print("list")

if [] :
  print("empty")
```

*list 안에 요소가 1개 이상 있다면 Ture, 비어있다면 False.

```python
if True :
  print("ture")

if False :
  print("false")
```

```python
num = 3
if num%3 :
  print(3)

if num%2 :
  print(2)
```

*3을 3으로 나눈 나머지는 0인데, 0은 False이므로 출력x.

```python
j = ".jpg"

if j in "cat.jpg" :
  print("CAT")

if j in "dog.png" :
  print("DOG")
```

**if-else문**

if문만을 사용하면 if문의 조건이 참일 때는 특정 동작을 수행할 수 있지만, if조건이 거짓일 때는 어떠한 동작을 수행할 수 없다.

if조건이 거짓일 때 특정 동작을 수행하게 하고 싶다면 if~else문을 사용하면 된다.

```python
n = 90
if n >= 80 :
  print("시험에 합격하셨습니다.")
  print("축하합니다.")
else :
  print("시험에 불합격하셨습니다.")
  print("다음에 다시 도전하십시오.")

print("프로그램을 종료합니다.")
```

# **직접해보기**

입력된 수가 짝수인지 판단하는 프로그램을 만드세요.

--------------------------------------------

숫자 --> 12

짝수입니다.

--------------------------------------------

```python
num = int(input("숫자 --> "))
if num%2 == 0 :
  print("짝수입니다.")
else :
  print("홀수입니다.")
```

# **직접해보기**

입력받은 수가 10 이상이면 10배를 한 결과를 출력하세요. 10 미만이면 아무런 연산도 하지 않습니다.

--------------------------------------------

정수를 입력하세요 20

20을 10배 한 결과는 200 입니다.

프로그램을 마칩니다,

--------------------------------------------

--------------------------------------------

정수를입력하세요 5

프로그램을 마칩니다.

--------------------------------------------

```python
num = int(input("정수를 입력하세요 "))
if num >= 10 :
  print(f"{num}을 10배 한 결과는 {num*10}입니다.")

print("프로그램을 마칩니다.")
```

# **직접해보기**

입력받은 수가 음수이면 해당 수를 양수로 바꾸어주는 프로그램을 만드세요.

--------------------------------------------

숫자 --> -5

결과는 5.0

--------------------------------------------

--------------------------------------------

숫자 --> 3.4

결과는 3.4

--------------------------------------------

```python
num = float(input("숫자 --> "))
if num < 0 :
  print(f"결과는 {-num}")
else :
  print(f"결과는 {num}")
```

***Another solution)***

```python
num = float(input("숫자 --> "))
if num < 0 :
  num*=(-1)

print(f"결과는 {num}")
```
